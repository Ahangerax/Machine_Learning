```python
import os

def rename_images(folder_path, label):
    # Supported image extensions
    valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.gif', '.webp')
    
    # Check if path exists
    if not os.path.exists(folder_path):
        print(f"Error: The folder '{folder_path}' does not exist.")
        return

    # Get a list of files and sort them to keep things orderly
    files = sorted(os.listdir(folder_path))
    
    count = 1
    for filename in files:
        # Check if file is an image based on extension
        if filename.lower().endswith(valid_extensions):
            file_ext = os.path.splitext(filename)[1]
            
            # Construct the new name: label + number + original extension
            new_name = f"{label}{count}{file_ext}"
            
            # Full paths for renaming
            old_path = os.path.join(folder_path, filename)
            new_path = os.path.join(folder_path, new_name)
            
            try:
                os.rename(old_path, new_path)
                print(f"Renamed: {filename} -> {new_name}")
                count += 1
            except Exception as e:
                print(f"Could not rename {filename}: {e}")

    print(f"\nFinished! Renamed {count - 1} images.")

# --- SET YOUR DETAILS HERE ---
target_folder = "./dataset/train/cats"  # Change this to your folder path
my_label = "cat"                       # Change this to "dog", "bird", etc.

rename_images(target_folder, my_label)
```
