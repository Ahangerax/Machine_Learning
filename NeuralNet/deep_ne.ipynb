```python
import numpy as np

def relu(Z):
    return np.maximum(0,Z)

def relu_derivative(Z):
    return (Z > 0).astype(float)

def sigmoid(Z):
    return 1/(1+ np.exp(-Z))

def sigmoid_derivative(A):
    return A*(1-A)
```

```python
class DeepNeuralNetwork:

    def __init__(self, layer_sizes, lr=0.01):
        np.random.seed(42)
        self.layer_sizes = layer_sizes
        self.lr = lr
        self.W = []
        self.b = []

        for i in range(len(layer_sizes)-1):
            W = np.random.randn(layer_sizes[i+1], layer_sizes[i]) * 0.01
            b = np.zeros((layer_sizes[i+1], 1))
            self.W.append(W)
            self.b.append(b)
    
    def forward(self, x):
        self.Z = []
        self.A = [x]

        for i in range(len(self.W)-1):
            Z = np.dot(self.W[i], self.A[i]) + self.b[i]
            A = relu(Z)
            self.Z.append(Z)
            self.A.append(A)

        # output layer
        Z = np.dot(self.W[-1], self.A[-1]) + self.b[-1]
        A = sigmoid(Z)
        self.Z.append(Z)
        self.A.append(A)
        return A

    def backward(self, y):
        dA = -2 * (y - self.A[-1])

        for i in reversed(range(len(self.W))):
            if i == len(self.W)-1:
                dZ = dA * sigmoid_derivative(self.A[i+1])
            else:
                dZ = dA * relu_derivative(self.Z[i])

            dW = np.dot(dZ, self.A[i].T)
            db = dZ
            dA = np.dot(self.W[i].T, dZ)

            self.W[i] -= self.lr * dW
            self.b[i] -= self.lr * db
    
    def train(self, X, Y, epochs=1000):
        for epoch in range(epochs):
            loss_total = 0

            for i in range(X.shape[0]):
                x = X[i].reshape(-1,1)
                y = Y[i].reshape(-1,1)
                output = self.forward(x)
                loss = (y - output)**2
                loss_total += loss
                self.backward(y)

            if epoch % 100 == 0:
                print(f"Epoch {epoch}, Loss {loss_total[0][0]:.4f}")
    
    def predict(self, X):
        preds = []
        for i in range(X.shape[0]):
            x = X[i].reshape(-1,1)
            output = self.forward(x)
            pred = 1 if output > 0.5 else 0
            preds.append(pred)
        return np.array(preds)
```

```python
layer_sizes = [30,16,8,4,1]
```

```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()

X = data.data
Y = data.target.reshape(-1,1)

X = X / np.max(X, axis=0)

layer_sizes = [30, 16, 8, 4, 1]

dnn = DeepNeuralNetwork(layer_sizes, lr=0.01)

dnn.train(X, Y, epochs=1000)

preds = dnn.predict(X)

accuracy = np.mean(preds == Y.flatten())

print("\nAccuracy:", accuracy)
```
