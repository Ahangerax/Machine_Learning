```python
import numpy as np 

from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
```

```python
X = data.data
Y = data.target

print(X.shape)
print(Y.shape)
```

```python
X = X/np.max(X, axis=0)
```

```python
Y =Y.reshape(-1,1)
```

```python
np.random.seed(42)

input_size = 30
hidden_size = 10
output_size = 1

W1 = np.random.randn(hidden_size, input_size) * 0.01
b1 = np.zeros((hidden_size,1))

W2 = np.random.randn(output_size, hidden_size) * 0.01
b2 = np.zeros((output_size,1))

lr = 0.01
epochs = 1000
```

```python
def relu(Z):
    return np.maximum(0,Z)

def relu_derivative(Z):
    return (Z > 0).astype(float)

def sigmoid(Z):
    return 1/(1+np.exp(-Z))
```

```python
for epoch in range(epochs):

    loss_total = 0

    for i in range(X.shape[0]):
        
        x = X[i].reshape(-1,1)
        y = Y[i].reshape(1,1)


        # forwoard
        Z1 = np.dot(W1,x) + b1
        A1 = relu(Z1)

        Z2 =np.dot(W2,A1) + b2
        A2 =sigmoid(Z2)

        loss = (y - A2)**2
        loss_total += loss

        # backward
        dA2 = -2*(y - A2)
        dZ2 = dA2*A2*(1-A2)

        dW2 = np.dot(dZ2, A1.T)
        db2 = dZ2

        dA1 = np.dot(W2.T,dZ2)
        dZ1 = dA1 * relu_derivative(Z1)

        dW1 = np.dot(dZ1, x.T)
        db1 = dZ1

        # update
        W1 = W1 - lr*dW1
        b1 = b1 - lr*db1

        W2 = W2 - lr*dW2
        b2 = b2 - lr*db2
    
    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss_total}")
```

```python
correct = 0

for i in range(X.shape[0]):

    x = X[i].reshape(-1, 1)

    Z1 = np.dot(W1, x) + b1
    A1 = relu(Z1)

    Z2 = np.dot(W2, A1) + b2
    A2 = sigmoid(Z2)

    prediction = 1 if A2 > 0.5 else 0

    if prediction == Y[i]:
        correct += 1

accuracy = correct / X.shape[0]

print("\nAccuracy:", accuracy)
```
