A dataset is divided into training and testing sets. Training data is used to train the machine learning model, while testing data is used to evaluate its accuracy and check for errors.

Overfitting occurs when a model performs very well on the training data but poorly on unseen testing data. Underfitting occurs when a model is too simple or lacks sufficient features, causing it to perform poorly on both training and testing data.

Features are the input variables in Machine Learning that help the model learn patterns. Good and relevant features improve prediction accuracy and reduce the risk of overfitting or underfitting.

Training error is the error a model makes on the training dataset, showing how well it has learned the data. Testing error is the error on unseen testing data, showing how well the model generalizes to new data.

Cross-validation is a technique where the training data is divided into multiple folds. The model is trained and validated on each fold to ensure consistent performance and reduce the risk of overfitting.

Hyperparameter tuning is the process of optimizing a model’s external settings (hyperparameters) to improve accuracy and reduce errors.

Model comparison is the process of evaluating multiple models on a dataset to assess their performance. Best model selection involves choosing the model that consistently provides the highest accuracy and lowest errors on training and testing data.

Data preprocessing is the process of cleaning and transforming raw data to make it suitable for Machine Learning, including handling missing values, outliers, and irrelevant features to ensure accurate predictions.

Common steps in data preprocessing include handling missing values, scaling or normalizing data, encoding categorical variables, detecting and treating outliers, and analyzing data distribution to make it suitable for Machine Learning.

Handling missing values is crucial in Machine Learning because missing data can cause the model to learn incomplete or incorrect patterns, leading to inaccurate predictions.

Feature scaling is important in Machine Learning to reduce computational cost, prevent features with large ranges from dominating the model, and improve convergence of learning algorithms.

Data transformation in Machine Learning is the process of converting raw data into a suitable format for the model, such as encoding categorical data into numerical values for learning, and converting predictions back to original form for human interpretation.

Outliers are data points that significantly differ from the rest of the dataset. They can bias the model and increase prediction errors. Outliers are handled by removing them if they are invalid or scaling them if they are valid to maintain accuracy.

Data normalization is the process of scaling features to a common range to ensure that the model learns efficiently and accurately, preventing features with larger ranges from dominating the learning process.

Jab dataset me bohot saare features ho aur overfitting ka risk ho → Ridge/Lasso use karte hain

Use-case → Handwriting recognition, customer segmentation

Support Vector Machine (SVM) is a classification algorithm that separates data using a hyperplane. For non-linear data, it maps the data to a higher dimension using kernel functions to make separation possible while maximizing the margin between classes.

Decision Tree is a Machine Learning algorithm that classifies data by recursively splitting it into branches based on feature conditions, using measures like entropy or gini impurity, until a final decision is reached.

Random Forest is an ensemble classification algorithm that combines multiple Decision Trees trained on random subsets of data and features. The final prediction is made using majority voting, making it more accurate and less prone to overfitting than a single Decision Tree.

Unit 0 – Basics & Data Types

Data types: Labeled vs Unlabeled

Importance: Models learn patterns and make predictions from data

Supervised vs Unsupervised:

Supervised → labeled → predict outcomes
Unsupervised → unlabeled → find patterns/clusters

Semi-supervised: Small labeled + large unlabeled → expensive labeling

Reinforcement Learning: Rewards/punishments improve model

ML vs AI: ML = subset of AI, backbone of AI

Deep Learning: Uses neural networks for complex patterns

Generative AI: Automatically generates content based on learned patterns

Unit 1 – Data Preprocessing

Data cleaning: Collect, remove impurities, fill missing values, scale, handle outliers

Feature scaling: Prevent model bias toward larger values

Categorical encoding: Label, One-Hot, Ordinal

Outliers: Bias model → handle or remove

Normalization: Standardize data → reduce errors

Unit 2 – Model Training

Training vs Test error:

Training → error on training data
Test → error on unseen data

Overfitting vs Underfitting:

Overfitting → too perfect on training, fails real-world
Underfitting → too simple, less features → poor predictions

Cross-validation: Split data multiple times → reliable model

Hyperparameter tuning: Adjust model settings for optimal performance

Unit 3 – Algorithms Basics

Linear Regression: Simple numeric prediction

Polynomial Regression: Higher powers → non-linear curves

Ridge & Lasso: Linear regression with regularization

Logistic Regression: Predict categories

KNN: Assign unknown features to nearest cluster using distance metrics

SVM: Separate non-linear data in higher dimension

Naive Bayes: Classify based on probabilities (categorical/binary)

Decision Tree: Split data using yes/no (boolean) → categories

Random Forest: Multiple decision trees → reduce overfitting → stable

Boosting: Sequential models → each fixes previous errors

Bagging: Parallel models → reduce variance

Unit 4 – Distance Metrics

Euclidean: Straight-line distance

Manhattan: Grid distance (horizontal + vertical moves)

Minkowski: General formula (p=1 → Manhattan, p=2 → Euclidean)

Hamming: Count differences in equal-length strings (binary/categorical)

Unit 5 – Tree Concepts

Entropy: Measures impurity → Decision Tree splits aim to reduce it

Gini Impurity: Alternative to entropy → simpler/faster

Information Gain: Choose split that maximizes reduction in impurity

Unit 6 – Ensemble Learning

Ensemble Learning: Combine multiple models → stronger, more accurate predictions

Bagging: Parallel → reduces variance (Random Forest example)

Boosting: Sequential → fixes errors → reduces bias + improves accuracy

Memory trick: Bagging = parallel; Boosting = sequential

Unit 7 – Model Evaluation & Improvement

Model comparison: Evaluate multiple models on same data

Best model selection: Choose model with best accuracy + generalization

Cross-validation: Multiple splits → reliable performance

Hyperparameter tuning: Adjust settings → best performance

Unit 8 – Automation & Practice

Kaggle: Datasets, competitions, community → practice ML skills
